```python
import pandas as pd
from pathlib import Path

PROCESSED_DIR = Path("../data/processed")

files = sorted(
    PROCESSED_DIR.glob("*-clean.parquet")
)

print("Files found:", len(files))

for file in files:
    print(file.name)
```

```python

dfs = []

for file in files:
    temp = pd.read_parquet(file)
    dfs.append(temp)

clean_df = pd.concat(
    dfs,
    ignore_index=True
)

print("Total rows:", len(clean_df))
print("Columns:", clean_df.columns.tolist())
```

```python
duplicate_pairs = clean_df.duplicated(
    subset=["src", "tgt"]
).sum()

print("Exact duplicate source-target pairs:",
      duplicate_pairs)
```

```python
translation_counts = (
    clean_df.groupby("src")["tgt"]
    .nunique()
)

multiple_translations = (
    translation_counts > 1
).sum()

print(
    "English sources with multiple Malayalam translations:",
    multiple_translations
)
```

```python
print("Unique English sources:", clean_df["src"].nunique())
```

```python
print(
    "Repeated English sources:",
    clean_df["src"].duplicated().sum()
)
```
